# Experiment 03_2 — Text Model Optimisation (TF-IDF & LinearSVC)

Goal: Find the best settings for the classic TF-IDF text model via cross-validation.

1. **Grid Search (3-Fold CV)**: Systematic evaluation of binary weighting and SVM regularisation ($C$) on the training split.
2. **Classifier Comparison**: Benchmark `LinearSVC` against `LogisticRegression` (SGD) and `ComplementNB`.
3. **Validation Verification**: Evaluate the final optimised model once on the shared 16984 validation products.


## 1. Load the data

```python
import sys
import time
from pathlib import Path

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import f1_score
from sklearn.model_selection import GridSearchCV, cross_val_score
from sklearn.naive_bayes import ComplementNB
from sklearn.pipeline import Pipeline
from sklearn.svm import LinearSVC

# Resolve repository root directory
PROJECT_ROOT             = Path.cwd().resolve()
while PROJECT_ROOT.name in {"03_text_fineTuning", "notebooks"}:
    PROJECT_ROOT         = PROJECT_ROOT.parent

DATA_PATH                = PROJECT_ROOT / "data" / "processed" / "text_train_processed.csv"
OUTPUT_DIR               = PROJECT_ROOT / "models" / "outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_STATE             = 42
CV_FOLDS                 = 3
N_JOBS                   = 4

data                     = pd.read_csv(DATA_PATH)
train                    = data[data["split"] == "train"]
validation               = data[data["split"] == "validation"]

X_train, y_train         = train["text_combined"], train["prdtypecode"]
X_validation, y_validation = validation["text_combined"], validation["prdtypecode"]

print("Project root:", PROJECT_ROOT)
print(f"Train       : {len(train):,}")
print(f"Validation  : {len(validation):,}")
```

## 2. Grid search on LinearSVC

The 03 notebook showed that using all features beats a 50000 limit by 1.2 points (CV on train). So we keep all features.

We test two settings together:
- Binary TF-IDF: count only whether a word appears, not how often (yes or no)
- C: how closely the SVM fits the training data. A small C gives a simpler model.

That gives 8 combinations, each scored with 3-fold CV on train.

```python
pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=2)),
    ("svm",   LinearSVC(random_state=RANDOM_STATE)),
])

param_grid = {
    "tfidf__binary": [False, True],
    "svm__C":        [0.1, 0.5, 1, 2],
}

start_time   = time.time()
grid         = GridSearchCV(pipeline, param_grid, cv=CV_FOLDS, scoring="f1_weighted", n_jobs=N_JOBS)
grid.fit(X_train, y_train)
print(f"Grid search time: {(time.time() - start_time) / 60:.1f} min")

grid_results = pd.DataFrame(grid.cv_results_)[
    ["param_tfidf__binary", "param_svm__C", "mean_test_score", "std_test_score"]
]
grid_results.columns = ["binary", "C", "cv_weighted_f1", "cv_std"]
display(
    grid_results.sort_values("cv_weighted_f1", ascending=False).rename(columns={
        "binary": "Binary", "cv_weighted_f1": "CV weighted F1", "cv_std": "CV spread",
    }).style.hide(axis="index").format(precision=4)
)
```

```python
# Rule: keep binary=True only if its gain is bigger than the CV spread
best_by_binary = grid_results.loc[grid_results.groupby("binary")["cv_weighted_f1"].idxmax()].set_index("binary")
binary_gain    = best_by_binary.loc[True, "cv_weighted_f1"] - best_by_binary.loc[False, "cv_weighted_f1"]
cv_spread      = grid_results["cv_std"].max()

use_binary     = bool(binary_gain > cv_spread)
best_C         = best_by_binary.loc[use_binary, "C"]

print(f"Binary gain: {binary_gain:+.4f}")
print(f"CV spread  : {cv_spread:.4f}")
print(f"Use binary : {use_binary}")
print(f"Best C     : {best_C}")

final_tfidf    = dict(ngram_range=(1, 2), min_df=2, binary=use_binary)
```

### Analysis

- C = 1 gives the best score. C = 0.5 and 2 are almost the same. C = 0.1 is about 2.5 points lower.
- Binary scores 0.8284 against 0.8235 without it. The gain (0.5 points) is larger than the CV spread (0.3), so we keep binary TF-IDF.
- Chosen settings: all features, binary TF-IDF, C = 1.

## 3. Other models

We keep the chosen TF-IDF settings and only change the classifier:
- Complement Naive Bayes: a fast and simple text model.
- Logistic Regression (trained with SGD): a linear model like the SVM that also gives probabilities.

Each model is scored with the same 3-fold CV on train.

```python
other_models = {
    "LinearSVC (chosen)"       : LinearSVC(C=best_C, random_state=RANDOM_STATE),
    "Complement Naive Bayes"   : ComplementNB(),
    "Logistic Regression (SGD)": SGDClassifier(loss="log_loss", alpha=1e-5, random_state=RANDOM_STATE),
}

model_rows = []
for name, model in other_models.items():
    pipe       = Pipeline([("tfidf", TfidfVectorizer(**final_tfidf)), ("model", model)])
    start_time = time.time()
    scores     = cross_val_score(pipe, X_train, y_train, cv=CV_FOLDS, scoring="f1_weighted", n_jobs=N_JOBS)
    model_rows.append({
        "model"         : name,
        "cv_weighted_f1": scores.mean(),
        "cv_std"        : scores.std(),
        "time_min"      : (time.time() - start_time) / 60,
    })

display(
    pd.DataFrame(model_rows).sort_values("cv_weighted_f1", ascending=False).rename(columns={
        "model": "Model", "cv_weighted_f1": "CV weighted F1", "cv_std": "CV spread", "time_min": "Time (min)",
    }).style.hide(axis="index").format(precision=4)
)
```

### Analysis

- LinearSVC has the best score (0.8284).
- Logistic Regression is 2 points lower (0.8084).
- Complement Naive Bayes is 7 points lower (0.7600).
- We keep LinearSVC.

## 4. Final score on validation

We train the baseline and the best model on the full training set, and score both once on validation.

```python
baseline  = Pipeline([
    ("tfidf", TfidfVectorizer(max_features=50_000, ngram_range=(1, 2), min_df=2, binary=True)),
    ("svm",   LinearSVC(random_state=RANDOM_STATE)),
])

optimised = Pipeline([
    ("tfidf", TfidfVectorizer(**final_tfidf)),
    ("svm",   LinearSVC(C=best_C, random_state=RANDOM_STATE)),
])

final_rows = []
for name, model in [("Baseline (03 notebook)", baseline), ("Optimised", optimised)]:
    model.fit(X_train, y_train)
    pred = model.predict(X_validation)
    final_rows.append({
        "model"      : name,
        "weighted_f1": f1_score(y_validation, pred, average="weighted"),
        "macro_f1"   : f1_score(y_validation, pred, average="macro"),
    })

display(
    pd.DataFrame(final_rows).rename(columns={
        "model": "Model", "weighted_f1": "Weighted F1", "macro_f1": "Macro F1",
    }).style.hide(axis="index").format(precision=4)
)
```

### Analysis

- The optimised model reaches 0.8436 weighted F1 on validation, against 0.8313 for the baseline (+1.2 points).
- Macro F1 goes from 0.8192 to 0.8330.
- The gain comes from using all features. Binary TF-IDF and C = 1 were already used in the baseline.

## Conclusion

Final text model: TF-IDF on words and word pairs (all features, binary) with LinearSVC (C = 1). Weighted F1 on validation: 0.8436.
